```python
import hashlib
import time
```

```python
def sha256_string(s: str) -> str:
    
    #Compute the SHA-256 hash of a string.
    
    return hashlib.sha256(s.encode("utf-8")).hexdigest()


def sha256_file(path: str) -> str:
    
    #Compute the SHA-256 hash of a file.

    h = hashlib.sha256()

    with open(path, "rb") as f:
        for chunk in iter(lambda: f.read(8192), b""):
            h.update(chunk)

    return h.hexdigest()
```

```python
# Compute the SHA-256 digest of a string
text = "Hello Blockchain"
print("String:", text)
print("SHA-256:", sha256_string(text))

print()
```

```python
# Part (c): Proof of Work 

def proof_of_work(message: str, n: int):

    prefix = "0" * n
    nonce = 0

    start = time.perf_counter()

    while True:

        digest = sha256_string(f"{message}{nonce}")

        if digest.startswith(prefix):

            end = time.perf_counter()

            return {
                "nonce": nonce,
                "digest": digest,
                "attempts": nonce + 1,
                "seconds": end - start
            }

        nonce += 1
```

```python
# For n=3

print("Proof-of-Work (n = 3)")
result = proof_of_work("Blockchain", 3)

for key, value in result.items():
    print(f"{key}: {value}")
```

```python
# For n=4
print("\nProof-of-Work (n = 4)")
result = proof_of_work("Blockchain", 4)

for key, value in result.items():
    print(f"{key}: {value}")
```
